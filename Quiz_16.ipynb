```python
data = {"age": ["20", "21", "19"],
        "gpa": ["3.5", "3.7", "3.9"],
        "enrolled": ["True", "False" , "True"]}
```

# 1.

The incorrect types is the age which it is suppose to be a int instead of a string, gpa is suppose to be a float instead of a string, and enrolled is supoose to be a Boolean instead of a string.

# 2.

```python
converted_age = [int(x) for x in data['age']]
print(converted_age)
```

```python
converted_gpa = [__builtins__.float(y) for y in data['gpa']]
print(converted_gpa)
```

```python
converted_boolean = [__builtins__.bool(x) for x in data['enrolled']]
print(converted_boolean)
```

# 3.

```python
average_gpa = sum(converted_gpa) / len(converted_gpa)
print(f"The average GPA is: {average_gpa:.2f}")
```
