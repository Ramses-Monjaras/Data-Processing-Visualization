```python
import pandas as pa

df = pa.read_csv('https://raw.githubusercontent.com/nurfnick/Data_Viz/main/Data_Sets/AB_NYC_2019.csv')
```

```python
print("Null counts before imputation:")
print(df.isnull().sum())

df['name'] = df['name'].fillna('No Name')

df['reviews_per_month'] = df['reviews_per_month'].fillna(0)

print("\nNull counts after imputation for 'name' and 'reviews_per_month':")
print(df[['name', 'reviews_per_month']].isnull().sum())
```
