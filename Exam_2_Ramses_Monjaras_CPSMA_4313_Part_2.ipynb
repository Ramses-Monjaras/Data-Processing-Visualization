```python
%%bigquery --project vizproject-488423
SELECT AVG(trip_miles) as avg_trip_miles, payment_type
FROM `bigquery-public-data.chicago_taxi_trips.taxi_trips`
GROUP BY payment_type
```

2. Avg trips miles: 4.048919 Payment type: Moblie

3. Avg trips miles: 3.747887 Payment type: Way2ride

4. Avg trips miles: 3.289322 Payment type: Dispute

5. Avg trips miles: 7.545511 Payment type: Prcard

```python
%%bigquery --project vizproject-488423
SELECT
  COUNT(*) as fares_above_average
FROM
  `bigquery-public-data.chicago_taxi_trips.taxi_trips`
WHERE
  fare > (
    SELECT
      AVG(fare)
    FROM
      `bigquery-public-data.chicago_taxi_trips.taxi_trips`
    WHERE
      fare > 0
  )
  AND fare > 0
```

The fares above average is 55515843
