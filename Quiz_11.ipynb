I'm going to use the usa_1910_current and usa_1910_2013 tables since they both have similar entries.

#INNER JOIN


```python
%%bigquery --project vizproject-488423
SELECT
    `current`.state,
    `current`.gender,
    `current`.year,
    `current`.name,
    `current`.number AS current_number,
    `old`.number AS old_number
FROM
    `bigquery-public-data.usa_names.usa_1910_current` AS `current`
INNER JOIN
    `bigquery-public-data.usa_names.usa_1910_2013` AS `old`
ON
    `current`.state = `old`.state AND
    `current`.gender = `old`.gender AND
    `current`.year = `old`.year AND
    `current`.name = `old`.name
LIMIT 5;
```

The inner join joins the state, gender, year, and name, the table shows:

0. state: AK, gender: F, year: 1985, name: Alice, current_number: 13, old_number: 13

1. state: AK, gender: F, year: 1918, name: Julia, current_number: 7, old_number: 7

2. state: AK, gender: F, year: 1920, name: Annie, current_number: 6, old_number: 6

3. state: AK, gender: F, year: 1921, name: Marget, current_number: 10, old_number: 10

4. state: AK, gender: F, year: 1925, name: Florence, current_number: 6, old_number: 6

#LEFT JOIN


```python
%%bigquery --project vizproject-488423
SELECT
    `current`.state,
    `current`.gender,
    `current`.year,
    `current`.name,
    `current`.number AS current_number,
    `old`.number AS old_number
FROM
    `bigquery-public-data.usa_names.usa_1910_current` AS `current`
LEFT JOIN
    `bigquery-public-data.usa_names.usa_1910_2013` AS `old`
ON
    `current`.state = `old`.state AND
    `current`.gender = `old`.gender AND
    `current`.year = `old`.year AND
    `current`.name = `old`.name
LIMIT 5;
```

The left join returns all the rows from the left table and only tje matched rows from the other table, which looks like this:

#RIGHT JOIN

```python
%%bigquery --project vizproject-488423
SELECT
    `current`.state,
    `current`.gender,
    `current`.year,
    `current`.name,
    `current`.number AS current_number,
    `old`.number AS old_number
FROM
    `bigquery-public-data.usa_names.usa_1910_current` AS `current`
RIGHT JOIN
    `bigquery-public-data.usa_names.usa_1910_2013` AS `old`
ON
    `current`.state = `old`.state AND
    `current`.gender = `old`.gender AND
    `current`.year = `old`.year AND
    `current`.name = `old`.name
LIMIT 5;
```

Right join retuns all entries from the right table and the matching entries from the left, which return this table:

#FULL OUTER JOIN

```python
%%bigquery --project vizproject-488423
SELECT
    `current`.state,
    `current`.gender,
    `current`.year,
    `current`.name,
    `current`.number AS current_number,
    `old`.number AS old_number
FROM
    `bigquery-public-data.usa_names.usa_1910_current` AS `current`
FULL OUTER JOIN
    `bigquery-public-data.usa_names.usa_1910_2013` AS `old`
ON
    `current`.state = `old`.state AND
    `current`.gender = `old`.gender AND
    `current`.year = `old`.year AND
    `current`.name = `old`.name
LIMIT 5;
```

The full outer join returns all rows from both left and right tables by combining the matching rows, which look like this:

0. state: AK, gender: F, year: 1910, name: Mary, current_number: 14, old_number: 14

1. state: AK, gender: F, year: 1910, name:Annie, current_number: 12, old_number: 12
