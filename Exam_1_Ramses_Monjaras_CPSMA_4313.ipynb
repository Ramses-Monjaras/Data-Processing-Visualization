# 1. Create Visualization

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/nurfnick/Data_Sets_For_Stats/refs/heads/master/CuratedDataSets/2022WorldCupPlayers.csv')
df
```

Flag.jpg

```python
sns.histplot(data=df, x= 'goals')
plt.title('Goals Distribution')
plt.xlabel('Goals')
plt.ylabel('Frequency')
plt.show()
```

I can see that this graph shows how rare it is to score a single goal. It also shows how impressive a player can be if they score more than once.

```python
sns.histplot(data=df, x = 'goals', hue = 'assists', palette= 'muted')
plt.title('Goals and Assists Distribution')
plt.xlabel('Goals and Assists')
plt.ylabel('Frequency')
plt.show()
```

I can see that this graph shows how common it is for a player not score. It also shows not there are not many players that are getting assists.

```python
sns.set_style("whitegrid")
sns.boxplot(data=df, x = 'team', y = 'cards_yellow')
plt.title('Yellow Cards Distribution by Nationality')
plt.xlabel('Nationality')
plt.ylabel('Yellow Cards')
plt.xticks(rotation=90)
plt.show()
```

This graph shows that Argentina has the extreme outlier of 3 yellow cards. This graph also shows that it is common to get 2 yellow card compared to just getting 1.

```python
df['is_brazil'] = df['team'] == 'Brazil'
sns.scatterplot(data=df, x= 'goals', y= 'assists', hue='is_brazil', palette={True: 'red', False: 'blue'}, s=100)
plt.title('Goals vs Assists (Brazil Highlighted)')
plt.xlabel('Goals')
plt.ylabel('Assists')
plt.legend(title='Brazil')
plt.show()
```

This scatter plot shows there is no correlation between goals and assists. The points are all over the place.

```python
sns.barplot(data=df, x= 'position', y= 'cards_yellow' )
plt.title('Yellow Cards Distribution by Position')
plt.xlabel('Position')
plt.ylabel('Yellow Cards')
plt.show()
```

I can see that the defender position receives more yellow cards compared to all the different positions. I am suprise to see that the goal keeper received any amount of yellow cards

```python
mean_score = 95
std_dev = 2

simulated_scores = np.random.normal(mean_score, std_dev, 1000)

plt.figure(figsize=(10, 6))
sns.histplot(simulated_scores, kde=True, color='green', bins=20)

plt.axvline(x=90, color='red', linestyle='--', label='Minimum "A" Score (90)')

plt.title('My Simulated Certainty Distribution for Getting an A')
plt.xlabel('Hypothetical Assignment Score (out of 100)')
plt.ylabel('Frequency / Density')
plt.legend()
plt.xlim(75, 100)
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

As you can see this graph starts around 90 telling you that I am certain I will at least a 90 on this assigment. However, this graph shows that I have a high chance of getting around a 95.
