```python
import pandas as pa
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

df = pa.read_csv('https://raw.githubusercontent.com/nurfnick/Data_Viz/main/Data_Sets/AB_NYC_2019.csv')
df
```

```python
plt.figure(figsize=(6, 4))
custom_colors = ['green', 'red', 'blue', 'purple', 'orange']
ax = sns.barplot(x='neighbourhood_group', y='price', hue='neighbourhood_group', data=df, palette=custom_colors, estimator=np.max)
for container in ax.containers:
    ax.bar_label(container)
plt.title('Maximum Price by Neighbourhood Group')
plt.xlabel('Neighbourhood Group')
plt.ylabel('Max Price')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.barplot(x='neighbourhood_group', y='price', hue='room_type', data=df, estimator=np.median)
plt.title('Average price by Neighbourhood Group and Room Type')
plt.xlabel('Neighbourhood Group')
plt.ylabel('Average price')
plt.legend(title='Room Type')
plt.show()
```
