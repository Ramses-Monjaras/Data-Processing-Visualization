```python
%%bigquery --project vizproject-488423

SELECT COUNT(*) as number_trips, start_station_name
FROM `bigquery-public-data.austin_bikeshare.bikeshare_trips`
GROUP BY start_station_name
ORDER BY number_trips DESC
LIMIT 1 OFFSET 1
```

The second most popular strating station is the 21st & Speedway @PCL with 71145 trips.

```python
%%bigquery --project vizproject-488423

SELECT COUNT(*) as round_trips_over_hour
FROM `bigquery-public-data.austin_bikeshare.bikeshare_trips`
WHERE start_station_name = end_station_name AND duration_minutes > 60
```

The number of trips that lasted over an hour and were a round trip is 70666.
