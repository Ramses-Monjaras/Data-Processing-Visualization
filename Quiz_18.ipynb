```python
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/nurfnick/Data_Viz/main/Data_Sets/IOT-temp.csv')
df.head()
```

```python
df['noted_date'] = pd.to_datetime(df['noted_date'], format='%d-%m-%Y %H:%M')
df.info()
```

```python
min_date = df['noted_date'].min()
max_date = df['noted_date'].max()
print(f"Minimum Date and Time: {min_date}")
print(f"Maximum Date and Time: {max_date}")
```

```python
df_sorted = df.sort_values(by='noted_date')
delta = df_sorted['noted_date'].diff().max()
print(f"The largest interval between temperature checks is: {delta}")
```

```python
average_datetime = df['noted_date'].mean()
print(f"The average datetime is: {average_datetime}")
```

```python
specified_date = '09-11-2018'
outside_temp_on_date = df[(df['noted_date'].dt.date == pd.to_datetime(specified_date).date()) & (df['out/in'] == 'Out')]

average_outside_temp = outside_temp_on_date['temp'].mean()

print(f"The average outside temperature on {specified_date} was: {average_outside_temp:.2f}")
```
