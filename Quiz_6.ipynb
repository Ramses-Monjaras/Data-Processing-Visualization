```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/nurfnick/Data_Viz/refs/heads/main/Data_Sets/Activity_Dataset_V1.csv')
df
```

```python
plt.figure(figsize=(15,10))
sns.set_style("darkgrid")
ax = sns.scatterplot(x="time", y='distance',hue='workout_type',size = 'calories', data=df, alpha =0.6)
plt.title('Scatter Plot of Distance vs. Time')
plt.xlabel('Time')
plt.ylabel('Distance')
plt.legend(bbox_to_anchor=(1.05, 1), loc=2, borderaxespad=0.)
plt.show()
```
