```python
%%bigquery --project vizproject-488423
SELECT
  COUNT(*) as num_trips,
  AVG(duration_minutes) as avg_duration_minutes
FROM
  `bigquery-public-data.austin_bikeshare.bikeshare_trips`
WHERE
  start_station_name IS NOT NULL
GROUP BY
  start_station_name
ORDER BY
  num_trips DESC
LIMIT 5
```

The average time for a trip based on starting points are:
1. 21st/Speedway is 15.45 minutes

2. 21st & Speedway is 19.92 minutes

3. Dean Keeton/Speedway is 15.75 minutes

4. Zilker Park is 43.59 minutes

5. 26th/Nueces is 16.10 minutes
