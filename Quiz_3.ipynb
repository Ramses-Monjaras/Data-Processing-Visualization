##Load Libraries

```python
import numpy as np
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns
```

##Load dataset

```python
df =pd.read_csv('https://raw.githubusercontent.com/nurfnick/Data_Viz/main/Data_Sets/iris.csv')
df
```

## Bar Chart

```python
sns.set_style("dark")
plt.figure(figsize=(10,5))
sns.barplot(data=df, x='Class', y='SepalWidth', hue='Class', palette='dark')
plt.title('Bar Plot of Sepal Width by Class')
plt
```

##Created visualization of my favorite variable

```python
sns.set_style("whitegrid")
plt.figure(figsize=(10,5))
sns.scatterplot(data=df, x='SepalLength', y='SepalWidth', hue='Class', palette='bright')
plt.title('Scatter Plot of Sepal Length vs Sepal Width')
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.show()
plt
```

## Histogram

```python
sns.set_style("darkgrid")
plt.figure(figsize=(10,5))
sns.histplot(data=df, x='PedalLength', kde=True, hue='Class', palette='muted')
plt.title('Histogram of Pedal Length')
plt.xlabel('Pedal Length')
plt.ylabel('Frequency')
plt.show()
plt
```

##Box Plot

```python
sns.set_style("whitegrid")
plt.figure(figsize=(10,5))
sns.boxplot(data=df, x='Class', y='SepalLength', hue='Class', palette='Paired')
plt.title('Box Plot of Sepal Length by Class')
plt.xlabel('Class')
plt.ylabel('Sepal Length')
plt.show()
plt
```

## Violin

```python
sns.set_style("whitegrid")
plt.figure(figsize=(10,5))
sns.violinplot(data=df, x='Class', y='PedalWidth', hue='Class', palette='colorblind')
plt.title('Violin Plot of Pedal Width by Class')
plt.xlabel('Class')
plt.ylabel('Pedal Width')
plt.show()
plt
```
